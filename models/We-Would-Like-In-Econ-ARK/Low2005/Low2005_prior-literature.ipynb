## Prior Literature

### Overview

{cite:t}`low2005` builds on the precautionary-saving literature in incomplete markets, where households smooth shocks using assets when insurance is limited. Classic contributions such as {cite:t}`deaton1991`, {cite:t}`carroll1994`, and {cite:t}`gourinchas2002` established how uncertainty and borrowing constraints shape consumption and wealth over the life cycle. Parallel labor-supply research ({cite:t}`heckman1974`; {cite:t}`blundell1993`; {cite:t}`blundell1998`) emphasized intertemporal substitution and the joint determination of work and consumption.

Within this context, {cite:t}`low2005` integrates flexible labor supply into a lifecycle precautionary-saving framework, showing that labor adjustments and savings are jointly chosen insurance margins.

### Key Foundational Papers

- {cite:t}`deaton1991` — A core benchmark for consumption-smoothing behavior under liquidity constraints and idiosyncratic risk.

- {cite:t}`carroll1994` — Quantified how expected income risk affects current consumption and precautionary behavior.

- {cite:t}`gourinchas2002` — Developed lifecycle evidence and structural interpretation of consumption dynamics over age.

- {cite:t}`heckman1974` — Linked lifecycle consumption and labor supply choices in an intertemporal framework.

- {cite:t}`blundell1993` — Provided empirical evidence on labor-supply intertemporal substitution, central for models with endogenous hours.

### The Gap This Paper Addresses

Many earlier models of precautionary savings abstracted from endogenous labor supply ({cite:t}`deaton1991`; {cite:t}`carroll1994`), while labor-supply studies often did not fully embed incomplete-markets precautionary motives. {cite:t}`low2005` fills this gap by jointly modeling consumption, hours, and asset accumulation across the life cycle under wage uncertainty. The paper shows how allowing labor adjustment changes both the level and age profile of saving, helping reconcile model outcomes with observed lifecycle patterns.
