## Hamish W. Low

## Self-Insurance in a Life-Cycle Model of Labour Supply and Savings (RED 2005)

#### Summary

This paper studies how allowing flexible labor supply changes precautionary saving and consumption dynamics in a life-cycle model with incomplete markets.

The main findings are listed below:

- Labor-supply flexibility changes asset profiles: more borrowing when young and more saving in middle age relative to fixed-hours models.
- Income uncertainty induces households to work more and consume less when young, increasing precautionary asset accumulation.
- With flexible hours, the utility cost of accumulating precautionary assets is lower, so equilibrium saving is typically higher.
- Jointly modeling labor supply and consumption better matches observed life-cycle profiles of hours and consumption.

### Non-Technical Methodical Overview

Many classic precautionary-saving models assume fixed labor supply. This paper shows that this misses an important margin of self-insurance.

Mechanism:

1. Households can respond to future wage risk by working harder early in life.
2. Extra early-life labor income allows larger buffer-stock savings.
3. Later in life, as uncertainty resolves and assets are higher, labor supply can decline.

The paper uses calibrated simulations to compare certainty vs uncertainty and flexible-hours vs fixed-hours economies.

### The Model

The paper formulates a finite-horizon life-cycle problem where agents choose consumption and leisure each period under uninsurable wage risk.

#### Dynamic program (paper equations (1)-(3))

$$
\max_{\{c_s,\ell_s\}}\;E_t\left[\sum_{s=t}^{T}\beta^{s-t}u(c_s,\ell_s)\right]
$$

subject to

$$
A_{s+1}=(1+r)\left[A_s+(H-\ell_s)w_s-c_s\right],
$$

with terminal condition

$$
A_{T+1}=0,
$$

and feasibility for leisure

$$
H \ge \ell_s \ge 0.
$$

#### Wage process (paper equations (4)-(5))

$$
\ln w_t = \ln w_0 + \alpha_1 t + \alpha_2 t^2 + \ln w_t^P + \nu_t, \qquad \nu_t \sim N\!\left(-\frac{\sigma_\nu^2}{2},\sigma_\nu^2\right),
$$

$$
\ln w_t^P = \ln w_{t-1}^P + \varepsilon_t, \qquad \varepsilon_t \sim N\!\left(-\frac{\sigma_\varepsilon^2}{2},\sigma_\varepsilon^2\right).
$$

So wages combine a deterministic age profile, a permanent random-walk component, and a transitory shock.

#### First-order conditions (paper equations (6)-(7))

$$
\frac{\partial u(c_t,\ell_t)}{\partial c_t}=\beta(1+r)E_t\left[\frac{\partial u(c_{t+1},\ell_{t+1})}{\partial c_{t+1}}\right],
$$

$$
\frac{\partial u(c_t,\ell_t)}{\partial \ell_t}=w_t\frac{\partial u(c_t,\ell_t)}{\partial c_t}\quad(\ell_t>0).
$$

These conditions jointly pin down intertemporal consumption smoothing and intratemporal labor-leisure trade-offs under uncertainty.

#### Quantitative strategy

The paper solves nonlinear policy functions numerically and calibrates to lifecycle moments (hours and wealth), then compares certainty vs uncertainty and flexible-hours vs fixed-hours economies.

### Results

1. **Flexibility under certainty:** Hours track wages more closely; borrowing/saving profiles become more hump-shaped.
2. **Adding uncertainty:** Households shift effort to younger ages, consume less early on, and accumulate more buffer assets.
3. **Flexibility under uncertainty:** The pre-emptive labor-supply response dominates, increasing desired precautionary wealth relative to fixed-hours benchmarks.
4. **Preference sensitivity:** Main qualitative conclusions are robust across alternative preference specifications, though magnitudes vary.

Overall, the model can reproduce key empirical life-cycle patterns better than fixed-labor alternatives.

### Conclusion

Labor-supply flexibility is a central self-insurance margin. Ignoring it can bias inference about precautionary motives, savings behavior, and consumption growth over the life cycle. The paper demonstrates that uncertainty and labor-supply choice jointly shape realistic age profiles of labor, consumption, and assets.

### Limitations

- The environment is partial equilibrium with a simplified asset structure.
- Participation margins, fixed work costs, and richer institutional labor frictions are abstracted from.
- Retirement and pension structure are stylized.
- Quantitative outcomes depend on assumed wage-process and preference calibration choices.
