## Subsequent Literature

### Research Directions Since Heathcote, Storesletten, and Violante (2014)

Subsequent research has expanded the partial-insurance framework in several directions. A first strand deepens the connection between micro heterogeneity and macro outcomes, including continuous-time and HANK-style formulations ({cite:t}`achdou2021`; {cite:t}`kaplan2018`).

A second strand focuses on labor-market structure and inequality decomposition under richer technology and task change environments ({cite:t}`autor2019`; {cite:t}`erosa2016`; {cite:t}`guvenen2023`). These papers continue to emphasize the joint role of wages, hours, and insurance channels.

A third strand studies policy design under partial insurance, extending analytical insights to taxation, redistribution, and welfare metrics ({cite:t}`heathcote2017`; {cite:t}`heathcote2021`; {cite:t}`nardi2020`).

### Key Subsequent Papers

- {cite:t}`heathcote2017` — Develops an analytical optimal-tax framework that builds directly on partial-insurance logic.

- {cite:t}`achdou2021` — Provides tractable continuous-time methods for income/wealth distributions that extend heterogeneous-agent insurance analysis.

- {cite:t}`kaplan2018` — Clarifies how micro heterogeneity shapes macro transmission, reinforcing the importance of distributional state variables.

- {cite:t}`autor2019` — Documents concentration and polarization patterns that motivate richer labor-risk structures in quantitative models.

- {cite:t}`guvenen2023` — Uses administrative data and model-based decomposition to revisit risk, inequality, and mobility channels.

### Open Questions and Gaps

1. **Insurance measurement across datasets.** Different data sources (survey vs. administrative) can imply different risk-sharing estimates, and reconciliation remains ongoing ({cite:t}`guvenen2023`; {cite:t}`trapeznikova2019`).

2. **Joint extensive-intensive labor margins.** Many models still emphasize hours responses, while participation and occupation transitions may be equally important ({cite:t}`sommer2016`; {cite:t}`erosa2016`).

3. **Policy robustness under high heterogeneity.** Optimal tax and transfer prescriptions can be sensitive to assumed preference and risk heterogeneity ({cite:t}`heathcote2021`; {cite:t}`nardi2020`).

4. **Macro-finance linkages.** Integrating partial insurance with asset-pricing and financial-frictions evidence remains an open frontier ({cite:t}`herskovic2015`; {cite:t}`acharya2020`).
