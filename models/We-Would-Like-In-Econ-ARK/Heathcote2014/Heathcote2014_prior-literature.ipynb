## Prior Literature

### Overview

{cite:t}`heathcote2010b` is part of a broad literature quantifying how idiosyncratic wage risk maps into consumption and labor-supply inequality. Empirical work on partial insurance, such as {cite:t}`blundell2008` and {cite:t}`krueger2006`, documented that consumption inequality rises less than earnings inequality, suggesting substantial but incomplete risk sharing. At the same time, lifecycle risk-sharing analyses like {cite:t}`storesletten2004` highlighted how insurance patterns vary with age and shock persistence.

On the theory side, heterogeneous-agent and incomplete-markets frameworks ({cite:t}`constantinides1996`; {cite:t}`cole2001`) provided key foundations, but often lacked transparent identification of insurance channels. {cite:t}`heathcote2010` advances this agenda by combining a tractable structural framework with moments from wages, hours, and consumption.

### Key Foundational Papers

- {cite:t}`blundell2008` — Estimated consumption insurance against permanent/transitory income shocks and set a benchmark pass-through decomposition.

- {cite:t}`krueger2006` — Documented the divergence between income and consumption inequality and emphasized the role of partial insurance.

- {cite:t}`storesletten2004` — Quantified lifecycle risk sharing and connected micro income risk to aggregate implications.

- {cite:t}`constantinides1996` — Provided a seminal heterogeneous-consumer asset-pricing framework under idiosyncratic risk.

- {cite:t}`cole2001` — Analyzed efficient allocations with hidden information/storage, informing interpretations of feasible risk-sharing arrangements.

### The Gap This Paper Addresses

Earlier studies typically either estimated reduced-form insurance coefficients or solved richer structural models with limited analytical transparency. {cite:t}`heathcote2010` addresses this gap by providing closed-form equilibrium relationships that jointly discipline wages, hours, and consumption moments, and by showing that labor-supply data can identify risk-sharing parameters even when consumption data are limited. This makes the framework both empirically implementable and theoretically interpretable.
