## Martin Floden

## The Effectiveness of Government Debt and Transfers as Insurance (JME 2001)

#### Summary

This paper studies how government debt and redistributive transfers provide insurance in an economy with incomplete markets and idiosyncratic income risk.

The main findings are listed below:

- Both public debt and transfers can improve household risk sharing when private insurance is limited.
- Transfers reduce inequality, while higher public debt tends to worsen equity by disproportionately benefiting wealthier asset holders.
- Debt can raise welfare when transfers are below their optimal level; however, debt's welfare gain largely disappears when transfers are already set optimally.
- Looking only at utilitarian welfare can hide large offsetting components: gains from improved insurance and losses from worsened distribution.

### Non-Technical Methodical Overview

The paper asks a policy-design question: if households face uninsurable income risk and borrowing constraints, should the government use debt, transfers, or both to improve outcomes?

The mechanism is intuitive:

1. **Transfers as direct insurance:** Taxes and transfers shift resources from temporarily high-income to low-income households.
2. **Debt as liquidity:** Higher debt can relax effective borrowing constraints and alter equilibrium prices, changing households' ability to smooth shocks.
3. **Distributional trade-offs:** Even if aggregate risk sharing improves, policy instruments may alter who gains and who loses.

To separate these channels, the paper decomposes welfare effects into (i) level/distortion effects, (ii) insurance effects, and (iii) redistribution effects.

### The Model

The paper uses a stationary balanced-growth heterogeneous-agent model with incomplete markets, no private insurance, and endogenous labor supply.

#### Production and prices (paper Section 2.1)

Aggregate production is written as

$$
Y = F(X,K,H)=K^{\theta}(XH)^{1-\theta},
$$

with competitive factor prices

$$
r=F_K(X,K,H)-\delta,\qquad w=F_H(X,K,H).
$$

#### Government budget (paper Section 2.2)

The government chooses debt and transfers, with budget accounting of the form

$$
G + B + rD = D' - D + T,
$$

where $T$ is tax revenue and $D$ denotes public debt.

#### Household problem in detrended variables (paper equations (4)-(6))

Using detrended notation $(\hat a,\hat c)$, the budget constraint is

$$
(1+g)\hat a' = \hat b + \bar w z h + (1+\bar r)\hat a - \hat c,
$$

with borrowing constraint

$$
\hat a'\ge 0.
$$

The Bellman equation is

$$
v(\hat a,s)=\max_{\hat c,h,\hat a'}\left\{u(\hat c,h)+\beta(1+g)^{1-\nu}\sum_{s'}\pi(s,s')v(\hat a',s')\right\}.
$$

#### Idiosyncratic productivity process (paper Section 2.5)

Labor efficiency combines permanent and transitory components and follows a finite-state Markov process, which is calibrated to match wage persistence and cross-sectional dispersion.

#### Quantitative strategy

Given policy pair $(B,D)$, the paper solves for stationary equilibrium prices, policy rules, and invariant distributions, then compares welfare decompositions into level/distortion, insurance, and inequality components.

### Results

Key policy counterfactuals compare changes in debt and transfers.

1. **Increase transfers:** Better insurance and lower inequality, but with distortionary costs.
2. **Increase debt (holding transfers fixed):** Better insurance in some regions, but potentially more unequal wealth outcomes.
3. **Joint design matters:** Debt and transfers are not independent policy substitutes; the effect of one depends strongly on the level of the other.

A central quantitative message is that debt can be welfare improving only in environments where transfers are not already at their welfare-maximizing level.

### Conclusion

The paper shows that government debt and redistribution both function as insurance tools under incomplete markets, but they differ sharply in their distributional implications. Debt can be useful as a second-best instrument when transfers are insufficient, yet it is not a blanket substitute for well-designed transfer policy. Welfare analysis should explicitly separate insurance and distribution channels, not just report a single aggregate welfare number.

### Limitations

- The analysis focuses on long-run/stationary comparisons and abstracts from transition dynamics after policy changes.
- Aggregate uncertainty/business-cycle shocks are largely absent, so cyclically contingent fiscal policy is not evaluated.
- Results depend on specific assumptions about market incompleteness and borrowing constraints.
- Political economy and implementation frictions for real-world debt and transfer policy are outside the model.
