## Heathcote, Storesletten, and Violante

## Consumption and Labor Supply with Partial Insurance: An Analytical Framework (AER 2014)

#### Summary

This paper develops a tractable framework for measuring risk sharing when households face idiosyncratic wage shocks and can adjust both consumption and labor supply.

The main findings are listed below:

- About 40% of permanent wage shocks pass through to consumption, implying substantial but incomplete insurance.
- The privately insured share of wage risk rises through the 1970s and then remains broadly stable from the early 1980s onward.
- Life-cycle wage shocks explain around half of wage dispersion but a much smaller share of dispersion in consumption and hours.
- Labor supply moments are highly informative for identifying insurance, even when consumption data are limited.

### Non-Technical Methodical Overview

The paper combines two traditions: structural modeling of incomplete markets and reduced-form measurement of consumption pass-through.

Core idea:

1. Split wage shocks into an **insurable** component and an **uninsurable** component.
2. Let households smooth shocks via private arrangements, taxes/transfers, and labor-supply adjustments.
3. Use joint moments of wages, hours, and consumption to infer the amount of risk sharing.

Because the model has closed-form allocations, the authors derive interpretable formulas for variances/covariances and prove identification of key parameters.

### The Model

The paper sets up a Yaari-style perpetual-youth economy with endogenous hours, heterogeneous disutility of labor, and partial insurance across idiosyncratic wage components.

#### Preferences (paper equations (1)-(2))

Lifetime utility for a cohort-$b$ agent is

$$
E_b\sum_{t=b}^{\infty}(\beta\delta)^{t-b}u(c_t,h_t;\phi),
$$

with period utility

$$
u(c_t,h_t;\phi)=\frac{c_t^{1-\gamma}-1}{1-\gamma}-\exp(\phi)\frac{h_t^{1+\sigma}}{1+\sigma}.
$$

#### Wage-risk decomposition (paper equation (3) and following)

$$
\log w_t = \alpha_t + \varepsilon_t,
$$

with

$$
\alpha_t=\alpha_{t-1}+\omega_t,\qquad \varepsilon_t=\kappa_t+\theta_t,\qquad \kappa_t=\kappa_{t-1}+\eta_t.
$$

Hence wages contain island-level persistent risk ($\alpha_t$), permanent individual risk ($\kappa_t$), and transitory risk ($\theta_t$).

#### Tax/transfer mapping (paper equation (4))

Disposable earnings are

$$
\tilde y_t = \lambda y_t^{1-\tau},
$$

where $\tau$ captures progressivity.

#### Closed-form allocation system (paper Proposition 1, equations (7)-(8))

A central tractability result is that equilibrium allocations can be written in log-linear form:

$$
\log c_t(s_t)=-(1-\tau)\hat\phi+(1-\tau)\left(\frac{1+\hat\sigma}{\hat\sigma+\gamma}\right)\alpha_t + C_t^a,
$$

$$
\log h_t(s_t)=-\hat\phi+\left(\frac{1-\gamma}{\hat\sigma+\gamma}\right)\alpha_t+\frac{1}{\hat\sigma}\varepsilon_t+H_t^a.
$$

These equations generate identifying restrictions for observed wage-hours-consumption moments.

#### Quantitative strategy

The model is estimated on PSID wages/hours and CEX consumption moments (levels and differences) to recover insurance shares and shock variances over time.

### Results

1. **Consumption insurance magnitude:** Permanent wage-shock pass-through to consumption is around 0.4.
2. **Insurance decomposition:** A large share of smoothing comes from private risk sharing, with additional roles for progressive taxation and labor-supply adjustment.
3. **Time variation:** Risk sharing improves in the 1970s and then stabilizes, despite rising wage inequality.
4. **Inequality decomposition:** Initial heterogeneity is crucial for consumption/hours dispersion, while life-cycle shocks are more important for wage dispersion.

The estimated framework matches observed cross-sectional moments for wages, hours, and consumption reasonably well across age and time.

### Conclusion

The paper provides a unified and analytically transparent approach to quantify partial insurance. It shows that substantial private and public smoothing coexists with meaningful residual exposure to wage risk. Incorporating labor supply is essential, both economically and empirically, for identifying and interpreting the degree of household insurance.

### Limitations

- The tractable structure imposes strong functional-form and shock-process assumptions.
- Some insurance channels (family networks, informal transfers, richer asset menus) are summarized rather than modeled explicitly.
- Empirical moments can be sensitive to measurement issues in survey consumption/hours data.
- The framework emphasizes partial-equilibrium style insurance decomposition and may abstract from broader general-equilibrium policy feedbacks.
