## Prior Literature

### Overview

{cite:t}`floden2001` sits in the incomplete-markets tradition where households face idiosyncratic risk, borrowing constraints, and limited insurance opportunities. Foundational work by {cite:t}`aiyagari1994` and {cite:t}`huggett1993` established how precautionary saving and uninsured shocks shape equilibrium asset holdings and welfare. Related general-equilibrium analyses of debt and taxation, such as {cite:t}`aiyagari1998` and {cite:t}`aiyagari1995`, emphasized that fiscal instruments can improve insurance but may introduce efficiency costs.

The paper also connects to earlier arguments that public debt can improve liquidity ({cite:t}`woodford1990`) and that redistributive taxation can serve an insurance role ({cite:t}`varian1980`). In this context, {cite:t}`floden2001` jointly evaluates debt and transfers, highlighting their distinct effects on risk sharing, efficiency, and inequality.

### Key Foundational Papers

- {cite:t}`aiyagari1994` — Introduced the canonical heterogeneous-agent framework with uninsured idiosyncratic risk and borrowing constraints, showing how precautionary saving affects aggregate outcomes.

- {cite:t}`huggett1993` — Characterized equilibrium asset-pricing implications in incomplete-insurance economies and clarified distributional heterogeneity in savings behavior.

- {cite:t}`aiyagari1998` — Studied the optimal debt level in incomplete-markets economies, providing a direct benchmark for fiscal debt policy.

- {cite:t}`varian1980` — Framed redistribution as social insurance, motivating transfer policies as risk-sharing tools.

- {cite:t}`woodford1990` — Developed the idea that public debt can provide private liquidity, linking debt policy to insurance and smoothing channels.

### The Gap This Paper Addresses

Much of the prior literature studied either debt policy ({cite:t}`aiyagari1998`; {cite:t}`woodford1990`) or redistributive transfers/taxation ({cite:t}`varian1980`; {cite:t}`aiyagari1995`) in relative isolation. {cite:t}`floden2001` fills this gap by comparing both instruments within a unified quantitative environment and by decomposing welfare effects into risk-sharing gains, efficiency costs, and distributional consequences. This integrated perspective clarifies when debt is beneficial, when transfers dominate, and why policy rankings depend on existing transfer levels.
